# Principal Component Analysis (PCA) Implementation
Importing the required libraries.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

Loading the dataset.

```python
from sklearn.datasets import load_breast_cancer

cancer_dataset = load_breast_cancer()
cancer_dataset.keys()
```

```python
df = pd.DataFrame(cancer_dataset["data"], columns=cancer_dataset["feature_names"])
df.head()
```

Standardization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
scaled_data
```

Applying the PCA Algorithms.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
```

```python
data_pca = pca.fit_transform(scaled_data)
data_pca
```

```python
pca.explained_variance_
```

What if? We don't mention the number of components.

```python
pca1 = PCA()
data_pca1 = pca1.fit_transform(scaled_data)
data_pca1
```

```python
pca1.explained_variance_
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(data_pca[:,0], data_pca[:,1], c=cancer_dataset["target"])
plt.xlabel("First principal component")
plt.ylabel("second principal component")
```

In the above plotting, it's clearly visible that the PCA selected the features that has highest variance. It makes the 2 classes clearly separable compressing the Training data to 2 features only.
